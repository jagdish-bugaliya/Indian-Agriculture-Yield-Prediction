```python

import pandas as pd 
import numpy as np 
import matplotlib.pyplot as plt 
```

```python
# opening files in dataframe
df_1 = pd.read_csv('data_set_of_2001-10.csv')
df_2=pd.read_csv('data_set_of_2011-20.csv')
df_1.columns=df_1.columns.str.lower()
```

```python

df_2.rename(columns={
    'crop_name': 'crop',
    'year': 'crop_year'
}, inplace=True)
```

```python
# merging data of 2001 to 2020 
df = pd.concat([df_1,df_2],ignore_index=True)
df
```

```python

# crop_yield calculation using production/area
df['crop_yield']=df['production']/df['area']
```

```python

# droping production column
df.drop(['production'],axis=1,inplace=True)
df
```

```python

df.to_csv('final_data_for_web.csv')
```

```python
df.drop(['state_name','district_name'],axis=1,inplace=True)
```

```python

# Changing all crops name in lowercase
df['crop']=df['crop'].str.lower()
df
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
numerical_columns = ['area','temperature_2m_mean','precipitation_sum','relative_humidity_2m_mean','wind_speed_10m_mean','latitude','longitude']

for col in numerical_columns:  # Replace with your actual column names
    plt.figure(figsize=(8, 4))
    sns.boxplot(x=df[col])
    plt.title(f"Boxplot for {col}")
    plt.show()
```

```python
# Graphical analysis of crop_yield data with respect to independent features
for i in df.columns.drop(['season','crop','crop_yield']):
    plt.scatter(df[i],df['crop_yield'])
    plt.xlabel(i)
    plt.ylabel('crop_yield')
    plt.show()
```

```python
# Data analysis after applying logp1
for i in df.columns.drop(['crop_yield']):
    plt.scatter(df[i],df['crop_yield'])
    plt.xlabel(i)
    plt.ylabel('crop_yield')
    plt.show()
```
