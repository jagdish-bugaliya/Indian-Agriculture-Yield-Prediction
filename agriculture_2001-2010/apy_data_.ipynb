```python
import pandas as pd
df = pd.read_csv("apy.csv")
```

```python
df
```

```python
# droping data before 2001
df = df.drop(df[df["Crop_Year"]<2001].index)
```

```python
# droping data after 2010
df = df.drop(df[df["Crop_Year"]>2010].index)
```

```python
# droping data for very low production state
df = df.drop(df[df["Production"] =="Andaman and Nicobar Islands"].index)
```

```python
# droping data for small scale prodution
df=df.drop(df[df['Production']<=100].index)
```
